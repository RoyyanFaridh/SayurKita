```python
import json
import time
from deep_translator import GoogleTranslator

# 1. Buka file JSON yang masih berbahasa Inggris
file_input = 'data_karbon_sayurkita_rata_rata.json'
file_output = 'data_karbon_sayurkita_id.json'

with open(file_input, 'r', encoding='utf-8') as f:
    data = json.load(f)

translated_data = {}
translator = GoogleTranslator(source='en', target='id')

print(f"Memulai proses terjemahan untuk {len(data)} jenis makanan...")

# 2. Lakukan iterasi (looping) untuk setiap makanan
for eng_key, metrik_karbon in data.items():
    # Ubah "pepper_sweet_red_raw" menjadi "pepper sweet red raw" agar mudah diterjemahkan
    teks_bersih = eng_key.replace('_', ' ')

    try:
        # Proses translasi via Google Translate API
        terjemahan = translator.translate(teks_bersih)

        # Ubah kembali hasil "paprika merah manis mentah" menjadi "paprika_merah_manis_mentah"
        ind_key = terjemahan.lower().replace(' ', '_').replace(',', '')

        # Masukkan ke dictionary baru
        translated_data[ind_key] = metrik_karbon

        print(f"Sukses: '{eng_key}' -> '{ind_key}'")

        # Beri jeda sangat singkat agar tidak diblokir oleh Google karena spamming
        time.sleep(0.2)

    except Exception as e:
        print(f"Gagal menerjemahkan '{eng_key}': {e}")
        # Jika gagal, tetap masukkan pakai bahasa Inggris agar data tidak hilang
        translated_data[eng_key] = metrik_karbon

# 3. Simpan ke file JSON baru
with open(file_output, 'w', encoding='utf-8') as f:
    json.dump(translated_data, f, indent=4)

print(f"\nSelesai! File bahasa Indonesia telah disimpan sebagai '{file_output}'")
```
